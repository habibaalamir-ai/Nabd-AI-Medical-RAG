````python
# Cell 1: Create Dockerfile

dockerfile_content = """FROM python:3.10-slim

WORKDIR /app

# Install dependencies
COPY requirements.txt .
RUN pip install --no-cache-dir -r requirements.txt

# Copy project files
COPY . .

# Expose Streamlit port
EXPOSE 8501

# Command to run the dashboard
CMD ["streamlit", "run", "app_ui.py", "--server.port=8501", "--server.address=0.0.0.0"]
"""

with open("nabd-cognitive-engine/Dockerfile", "w", encoding="utf-8") as f:
    f.write(dockerfile_content)

print("✅ Dockerfile created successfully.")
````

````python
# Cell 2: Create .dockerignore

dockerignore_content = """__pycache__/
*.pyc
*.pyo
*.pyd
.pytest_cache/
.git/
.gitignore
*.zip
"""

with open("nabd-cognitive-engine/.dockerignore", "w", encoding="utf-8") as f:
    f.write(dockerignore_content)

print("✅ .dockerignore created successfully.")
````
